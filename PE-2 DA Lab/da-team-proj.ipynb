```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = {
    "Student": ["A", "B", "C", "D", "E", "F"],
    "Marks": [72, 85, 60, 90, 78, 66]
}

df = pd.DataFrame(data)
df
```

```python
topper = df.loc[df["Marks"].idxmax()]

print("Topper:")
print(topper)
```

```python
df["Marks"].describe()
```

```python
plt.figure()
sns.histplot(df["Marks"], bins=5)
plt.title("Distribution of Marks")
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure()
sns.boxplot(x=df["Marks"])
plt.title("Marks Distribution")
plt.show()
```

```python
cutoff = df["Marks"].mean()
print("Cutoff Marks:", cutoff)
```

```python
cutoff = df["Marks"].quantile(0.75)
print("Cutoff (75th percentile):", cutoff)
```

```python
cutoff = df["Marks"].mean()

selected_students = df[df["Marks"] >= cutoff]

print("Students above cutoff:")
print(selected_students)
```

```python
mean_marks = df["Marks"].mean()
print("Mean:", mean_marks)
```

```python
variance = df["Marks"].var()

print("Variance:", variance)
```

```python
std_dev = np.sqrt(variance)

print("Standard Deviation:", std_dev)
```

```python
cutoff = mean_marks + std_dev

print("Cutoff Marks:", cutoff)
```

```python
selected_students = df[df["Marks"] >= cutoff]

print("Students above cutoff:")
print(selected_students)
```
